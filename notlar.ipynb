 <h2  style="background-color:rgb(255, 50, 200);"   > Sözlük(Dictionaries) </h2>

<li><b>clear():</b>Sozluk elemanini siler.</li>
<li><b>copy():</b> Sozlugu kopyalar. </li>
<li><b>fromkeys():</b>Returns a dictionary with the specified keys and value</li>
<li><b>get():</b>Sozlukte belirli bir anahtara ait degeri yazdirir</li>
<li><b>items():</b>Returns a list containing a tuple for each key value pair</li>
<li><b>keys():</b>Sozlukteki anahtar kelimeleri yazdirir/li>
<li><b>pop():</b>Sozlukden eleman cikarir.</li>
<li><b>popitem():</b>Removes the last inserted key-value pair.</li>
<li><b>setdefault():</b>Returns the value of the specified key. If the key does not exist: insert the key, with the specified value</li>
<li><b>update():</b> Bir sozlugu digerine ekleyerek gunceller.</li>
<li><b>values():</b> Sozlukteki anahtar kelimelere karsilik degerleri verir.</li>

<h3 style="background-color:DodgerBlue;">iki setden biri digerini kapsama durumunu belirleme issuperset()   </h3>

```python
<h3 style="background-color:DodgerBlue;">iki setden biri digerine altkume durumunu belirleme issubset()   </h3>
```

```python
from numpy.lib.scimath import sqrt
a = 1; b = 2; c = 10 # polinom katsayilari
r1 = (-b + sqrt(b**2 - 4*a*c))/(2*a)
r2 = (-b - sqrt(b**2 - 4*a*c))/(2*a)
print("Denklemin cozumu: r1 = {:g}, r2 = {:g}".format(r1,r2))

a = 1; b = 4; c = 1 # polinom katsayilarini degistirelim
r1 = (-b + sqrt(b**2 - 4*a*c))/(2*a)
r2 = (-b - sqrt(b**2 - 4*a*c))/(2*a)
print("Denklemin cozumu: r1 = {:g}, r2 = {:g}".format(r1,r2))
```
