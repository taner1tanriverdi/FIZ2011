> Demet lemanlari demet listeye cevrilerek yapilir.

```python
meyveler = ("elma", "armut", "uzum", "kavun","incir", "kivi", "elma", "armut", "uzum", "kavun","incir", "kivi", "elma", "armut", "uzum", "kavun","incir", "kivi")
liste = list(meyveler)
liste[0]="muz"
meyveler = tuple(liste)
meyveler
```

```python
x = ("apple", "banana", "cherry")
y = list(x)
y[1] = "kiwi"
x = tuple(y)

print(x)
```

```python
meyveler = ("elma", "muz", "kivi")
(kirmizi,sari, yesil) = meyveler
print(sari)
print(kirmizi)
print(yesil)
```

```python
sayilar = ("bir", "iki", "uc")
(bir,iki,uc) = sayilar
print(bir)
```

## **Demetlerde Dongu**

## Demetleri Birlestime

```python
demet1 = ("bir", "dort", "bes", "yedi", "alti", "sekiz")
demet2 = (1,2,3, "a", "bir")
demet3 =demet1 + demet2
print(demet3)
```
