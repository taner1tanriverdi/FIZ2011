```python
import matplotlib.pyplot as plt
import numpy as np

# x aralığını belirle (örneğin -2'den 6'ya kadar 0.1 adımlarla)
x = np.arange(-2, 6, 0.1)

# y değerlerini hesapla
y = x**2 - 4*x + 3

# Grafik oluştur
plt.figure(figsize=(8, 6))
plt.plot(x, y, label='y = x² - 4x + 3', color='blue')

# Eksen limitleri (min - max)
plt.xlim(-3, 7)
plt.ylim(-5, 15)

# Eksen etiketleri
plt.xlabel("x ekseni")
plt.ylabel("y ekseni")
plt.title("Parabol Grafiği")

# Grid (ızgara) ekle
plt.grid(True, which='both', linestyle='--', linewidth=0.7, alpha=0.7)

# Minimum ve maksimum noktaları işaretleme
x_min = x[np.argmin(y)]
y_min = np.min(y)
plt.scatter(x_min, y_min, color='red', label='Minimum nokta')

# Göster
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)

plt.subplot(221)   # 2 satır, 2 sütun → 1. subplot
plt.plot(x, np.sin(x))
plt.title("sin(x)")

plt.subplot(222)   # 2×2 → 2. subplot
plt.plot(x, np.cos(x))
plt.title("cos(x)")

plt.subplot(223)   # 3. subplot
plt.plot(x, np.tan(x))
plt.title("tan(x)")

plt.subplot(224)   # 4. subplot
plt.plot(x, np.exp(+x))
plt.title("exp(-x)")

plt.tight_layout()
plt.show()
```
