 <h2  style="background-color:rgb(255, 50, 200);"   > Setler(Sets) </h2>

 <li>add()                        :Sete  eleman ekler.</li>
 <li>remove()                     :Setin istenilen elemanini cikarir.</li>
 <li>discard()                    :Belirli elemani cikarir.</li>
 <li>pop()                        :Setden eleman cikarir.</li>
 <li>clear()                      :Setin elemanini siler.</li>
 <li>difference()                 :Setler arasindaki farki yazdirir.</li>
 <li>difference_update()          :Setler arasindaki farki bulup, güncel sonucu yazdirir.</li>
 <li>intersection()               :İki setin arasindaki kesişimini yazdirir.</li>
 <li>intersection_update()        :İki setin arasindaki kesişimini guncelleyip yazdirir.</li>
 <li>isdisjoint()                 :İki setin ayrik olup olmama durumunun doğru veya yanlis oldugunu yazdirir.</li>
 <li>issubset()                   :Bir setin diger setin alt seti olup olamdiginin yazdirilmasi.</li>
 <li>issuperset()                 :Bir setin diger seti kapsama durumunu yazdirir.</li>
 <li>symmetric_difference()       :Her iki setde bulunan elemanlari birlestirir, ortak eleman hariç.</li>
 <li>symmetric_difference_update():Her iki setde bulunan elemani cikarir. Her iki sette olmayan elemanlari birlestirir.</li>
 <li>union()                      :Setleri birleştirir.</li>
 <li>update()                     :Bir seti digerine ekleyerek gunceller.</li>

<h3 style="background-color:DodgerBlue;">iki setden biri digerini kapsama durumunu belirleme issuperset()   </h3>

```python
x = {"f", "e", "d", "c", "b", "a"}
y = {"a", "b", "c"}
z = x.issuperset(y)
print(z)            #x seti y setini kapsar
```

```python
<h3 style="background-color:DodgerBlue;">iki setden biri digerine altkume durumunu belirleme issubset()   </h3>
```

```python
x = {"f", "e", "d", "c", "b", "a"}
y = {"a", "b", "c"}
z = y.issubset(x)
print(z)            #y seti x setinin altsetidir. Doğru cevabi verir.
print(x.issubset(y))#x seti y setinin altseti değildir. Yanlis cevabi verir.
```
