```python
import matplotlib.pyplot as plt
from numpy import arange, exp, radians, sin

x1 = radians(arange(90, 1530, 1))  # 0'dan 1440 dereceye (8*pi radyan, ~25.13 rad)
x2 = radians(arange(0, 1440, 1)) 
y1 = 0.25*sin(x1)
y1a = 0.25*sin(x2)
y2a = y1 + y1a
y2 = 4* exp(-x2)
y3 = y1 * y2
y4 = y1 + y2
plt.scatter(x1, y1, s=8, label="sin(x)")
plt.scatter(x2, y1a, s=8, label="sin(x)v1")
plt.scatter(x2, y2a, s=8, label="sin(x)v2a")
plt.scatter(x2, y2, s=8, label="exp(-x)")
plt.scatter(x2, y3, s=8, label="sin(x)*exp(-x)")
plt.scatter(x2, y4, s=8, label="sin(x)+exp(-x)")
plt.xlabel("x (radyan)")
plt.ylabel("değer")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Fiziksel parametreler
A = 1.0
w0 = 1.0  # Doğal açısal frekans (omega_0)
beta = 0.2 * w0  # Grafikteki sönüm katsayısı (beta = 0.2 * omega_0)
w1 = np.sqrt(w0**2 - beta**2)  # Denklem (3.38): Sönümlü açısal frekans

# Zaman dizisi (grafikteki gibi ~2.25 salınım periyodu için)
t = np.linspace(0, 2.25 * (2 * np.pi / w0), 1000)

# Eğriler
x_sonumsuz = A * np.cos(w0 * t)  # beta = 0 durumu
x_sonumlu = A * np.exp(-beta * t) * np.cos(w1 * t)  # beta = 0.2 * w0 durumu
zarf_ust = A * np.exp(-beta * t)  # +A * exp(-beta * t)
zarf_alt = -A * np.exp(-beta * t)  # -A * exp(-beta * t)

# Grafik çizimi
fig, ax = plt.subplots(figsize=(8, 4.5))

ax.plot(t, x_sonumsuz, "--", color="black", linewidth=1.2)
ax.plot(t, x_sonumlu, "-", color="black", linewidth=1.6)
ax.plot(t, zarf_ust, "--", color="black", linewidth=1.1)
ax.plot(t, zarf_alt, "--", color="black", linewidth=1.1)

# Eksen düzenlemeleri (ders kitabı stilinde)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["bottom"].set_position("zero")
ax.spines["left"].set_position("zero")
ax.set_xticks([])
ax.set_yticks([])

# Etiketler ve notasyonlar
ax.set_ylabel("Genlik", fontsize=11, labelpad=10)
ax.text(t[-1] * 1.02, 0, "$t$", fontsize=12, va="center")

# Eğri üzerindeki açıklamalar
ax.text(2.2, 0.72, r"$Ae^{-\beta t}$", fontsize=11)
ax.text(4.8, -0.45, r"$-Ae^{-\beta t}$", fontsize=11)
ax.text(6.6, 0.85, r"$\beta=0$", fontsize=11)
ax.text(1.8, -0.52, r"$\beta=0.2\omega_0$", fontsize=11)

plt.tight_layout()
plt.show()
```
