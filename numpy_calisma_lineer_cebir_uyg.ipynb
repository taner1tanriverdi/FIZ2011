### İki Bilinmeyenli Denklem

AX = B

\[
\begin{cases}
4x  + 3y = 20 \\
-5x + 9y = 26
\end{cases}
\]



```python
import numpy as np

m_list = [[4, 3], [-5, 9]]
A = np.array(m_list)
print(A)
print('A')
inv_A = np.linalg.inv(A)
print(inv_A)

#X = [[x],[y]]

B = np.array([20, 26])
X = np.linalg.inv(A).dot(B)


print(X)
```

### Üç Bilinmeyenli Denklem

\[
\begin{cases}
4x + 3y + 2z = 25 \\
-2x + 2y + 3z = -10 \\
3x - 5y + 2z = -4
\end{cases}
\]

AX = B
A$^-1$AX = B

```python
A = np.array([[4, 3, 2], [-2, 2, 3], [3, -5, 2]])
B = np.array([25, -10, -4])
X = np.linalg.inv(A).dot(B)

print(X)
```

```python
import numpy as np

alpha = np.radians(90) # 90 derece döndürelim
c, s = np.cos(alpha), np.sin(alpha)

# Saat yönü tersine matris
R = np.array([[c, -s], 
              [s,  c]])

v = np.array([1, 0]) # X ekseni üzerinde bir vektör
v_yeni = R @ v       # Matris çarpımı

print(v_yeni) # Çıktı yaklaşık [0, 1] olacaktır (Y eksenine döndü)
```

```python
import cmath

a = 1
b = 5
c = 6

# calculate the discriminant
d = (b**2) - (4*a*c)

# find two solutions
sol1 = (-b-cmath.sqrt(d))/(2*a)
sol2 = (-b+cmath.sqrt(d))/(2*a)

print('The solution are {0} and {1}'.format(sol1,sol2))
```
