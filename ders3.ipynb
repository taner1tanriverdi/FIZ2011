# Bitwise işlemleri

```python
a = 5   # 0101 (binary)
b = 3   # 0011 (binary)

print(a & b)   # 0101 & 0011 = 0001 -> 1
print(a | b)   # 0101 | 0011 = 0111 -> 7
print(a ^ b)   # 0101 ^ 0011 = 0110 -> 6
print(~a)      # ~0101 -> ...1010 -> -6 (işaretli tam sayı)
print(a << 1)  # 0101 << 1 = 1010 -> 10 (sola kaydırma)
print(a >> 1)  # 0101 >> 1 = 0010 -> 2 (sağa kaydırma)
```

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.arange(0,5)
y = x**3-6*x**2+11*x-6
#plt.plot(x,y)
plt.scatter(x,y)
#plt.pie(x,y)
plt.xlabel("x-ekseni")
plt.ylabel("y-ekseni")
plt.title("parabol")
plt.show()

import matplotlib.pyplot as plt
import numpy as np

# x aralığını belirle (örneğin -2'den 6'ya kadar 0.1 adımlarla)
x = np.arange(-2, 6, 0.1)

# y değerlerini hesapla
y = x**3-6*x**2+11*x-6

# Grafik oluştur
plt.figure(figsize=(8, 6))
plt.plot(x, y, label='y = x² - 4x + 3', color='blue')

# Eksen limitleri (min - max)
plt.xlim(-3, 7)
plt.ylim(-5, 15)

# Eksen etiketleri
plt.xlabel("x ekseni")
plt.ylabel("y ekseni")
plt.title("Parabol Grafiği")

# Grid (ızgara) ekle
plt.grid(True, which='both', linestyle='--', linewidth=0.7, alpha=0.7)

# Minimum ve maksimum noktaları işaretleme
x_min = x[np.argmin(y)]
y_min = np.min(y)
plt.scatter(x_min, y_min, color='red', label='Minimum nokta')

# Göster
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# x aralığını belirle (örneğin -2'den 6'ya kadar 0.1 adımlarla)
x = np.arange(-2, 6, 0.1)

# y değerlerini hesapla
y = x**2 - 4*x + 3

# Grafik oluştur
plt.figure(figsize=(8, 6))
plt.plot(x, y, label='y = x² - 4x + 3', color='blue')

# Eksen limitleri (min - max)
plt.xlim(-3, 7)
plt.ylim(-5, 15)

# Eksen etiketleri
plt.xlabel("x ekseni")
plt.ylabel("y ekseni")
plt.title("Parabol Grafiği")

# Grid (ızgara) ekle
plt.grid(True, which='both', linestyle='--', linewidth=0.7, alpha=0.7)

# Minimum ve maksimum noktaları işaretleme
x_min = x[np.argmin(y)]
y_min = np.min(y)
plt.scatter(x_min, y_min, color='red', label='Minimum nokta')

# Göster
plt.legend()
plt.show()
```
